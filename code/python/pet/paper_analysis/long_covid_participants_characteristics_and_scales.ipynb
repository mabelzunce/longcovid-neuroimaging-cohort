```python
import pandas as pd
import os
print(os.getcwd())
pet_data = pd.read_csv("../../../../data/pet/pet_scans_info_covid_controls_ceunim_merged.csv")
study_data = pd.read_csv("../../../../data/study_summary.csv")
output_dir = "../../../../results/pet/paper_analysis/"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

group_col = "group"  # adjust if the group column has a different name
groups = ["LC", "Control"]
subjects_to_exclude = ['CP0106']  # add any subjects to exclude from the analysis here
df = pet_data[pet_data[group_col].isin(groups)]
df = df[~df["subject_id"].isin(subjects_to_exclude)]
# Get the subject_ids of the long covid participants
long_covid_subjects = df.loc[df[group_col] == "LC", "subject_id"].tolist()
# print the subject and count of long covid participants
print("Long COVID subjects:", long_covid_subjects)
print("Number of Long COVID subjects:", len(long_covid_subjects))


# Cols of the study_summary csv: ID,Grupo,FechaCognitivo,Edad,Genero,Altura ,Peso,BMI,Vacunado,NumVacunas,NumVacunasPreInfeccion,FechaUltimaDosis,DiagnosticoCovid,NumDiagnosticoCovid,FechaSintomasPrimeraVez,NumDosisPrePrimerInfeccion,DiasEntrePrimeraInfeccionYCuestionario,FechaSintomasUltimaVez,Internacion,CategoríaCOVID,Recuperado,DiagnosticoCOVIDProlongad,DolorDeCabeza,Fatiga,Olfato,Gusto,Disnea,DebilidadMuscular,DolorMuscular,Confusion,Comunicacion,Dormir,Memoria ,Atencion,Otro,PresionAlta,Diabetes,Asma,ColesterolAlto,InfartoCardiaco,AnginaPecho,EmboliaTrombosis,MedicacionPresionArterial,Aspirinas,Antiagregantes,Otro,CualOtro,Fumador,EQ-5D-5L,EQ-VAS,FAS,CategoriaFAS,IPAQ,CategoriaIPAQ,PSQI,CategoriaPSQI,NivelEducativo,Ocupacion,DiasEntrePrimeraInfeccionYCuestionario ,DiasEntreUltimaInfeccionYCuestionario ,TMT-A,TMT-A_perc,TMT-A-rendimiento,TMT-B,TMT-B_perc,TMT-B-rendimiento,WMS-R-DIR,WMS-R-DIR_perc,WMS-R-DIR-rendimiento,WMS-R-INV,WMS-R-INV_perc,WMS-R-INV-rendimiento,STROOP_P,STROOP_P_perc,STROOP_P-rendimiento,STROOP_C,STROOP_C_perc,STROOP_C-rendimiento,STROOP_P/C,STROOP_P/C_perc,STROOP_P/C-rendimiento,STROOP_P/C_INTERF,STROOP_P/C_INTERF_perc,STROOP_P/C_INTERF-rendimiento,MOCA,MOCA_perc,MOCA-rendimiento,HADSDepresion,HADSDepresionRendimiento,HADSAnsiedad,HADSAnsiedadRendimiento, PHQ-9, PHQ-9-rendimiento,GAD7,GAD7-rendimiento,DepresionEscalaUnificada,DepresionCategoriaUnificada,AnsiedadEscalaUnificada,AnsiedadCategoriaUnificada,UNSAM,Cortisol (pelo) pg/mg,M6a,BDNF,NFL,IndiceSocioeconomico,Indice de sintomas cognitivos,MRI,PET,WMHTotal,WMHPeriventricular,WMHDeep,TotalGM_B,TotalGM_L,TotalGM_R,WMV,sGMV,Ventricles,GMV
# Now from the study summary, get a separate df with only yhe covid participant from the PET data
study_covid_df = study_data[study_data["ID"].isin(long_covid_subjects)]
print("Study summary for Long COVID participants:")
display(study_covid_df)


```

```python
import numpy as np
from scipy import stats

# Define the three groups
covid_all = study_data[(study_data["Grupo"] == "COVID") & (~study_data["ID"].isin(subjects_to_exclude))]
control_all = study_data[(study_data["Grupo"] == "CONTROL") & (~study_data["ID"].isin(subjects_to_exclude))]
covid_pet = study_covid_df[~study_covid_df["ID"].isin(subjects_to_exclude)]

group_dfs = {"COVID": covid_all, "CONTROL": control_all, "COVID_PET": covid_pet}

# Resolve column names robustly (handles trailing spaces / accents)
def find_col(name):
    norm = lambda s: str(s).strip().lower().replace("í", "i").replace("\xad", "")
    for c in study_data.columns:
        if norm(c) == norm(name):
            return c
    return None

numeric_vars = [
    "DiasEntrePrimeraInfeccionYCuestionario",
    "FAS", "MOCA", "MOCA_perc", "DepresionEscalaUnificada", "AnsiedadEscalaUnificada",
]
categorical_vars = [
    "NumDosisPrePrimerInfeccion", "Fatiga", "Olfato", "Gusto", "Disnea",
    "DebilidadMuscular", "DolorMuscular", "Confusion", "Comunicacion", "Dormir",
    "Memoria", "Atencion", "Internacion", "CategoríaCOVID",
    "DepresionCategoriaUnificada", "AnsiedadCategoriaUnificada",
    "Indice de sintomas cognitivos", "CategoriaFAS", "PHQ-9-rendimiento", "GAD7-rendimiento",
]

results = []

# Numeric variables: descriptive stats + Kruskal-Wallis + pairwise Mann-Whitney
for var in numeric_vars:
    col = find_col(var)
    if col is None:
        print(f"Column not found: {var}")
        continue
    data = {g: pd.to_numeric(d[col], errors="coerce").dropna() for g, d in group_dfs.items()}
    row = {"variable": col}
    for g, s in data.items():
        row[f"{g} (n)"] = len(s)
        row[f"{g} median [IQR]"] = (f"{s.median():.2f} [{s.quantile(.25):.2f}-{s.quantile(.75):.2f}]"
                                    if len(s) else "NA")
        row[f"{g} mean±sd"] = f"{s.mean():.2f}±{s.std():.2f}" if len(s) else "NA"
    valid = [s for s in data.values() if len(s) > 0]
    try:
        row["Kruskal p"] = stats.kruskal(*valid).pvalue if len(valid) > 1 else np.nan
    except ValueError:
        row["Kruskal p"] = np.nan
    names = list(data.keys())
    for i in range(len(names)):
        for j in range(i + 1, len(names)):
            a, b = data[names[i]], data[names[j]]
            try:
                p = stats.mannwhitneyu(a, b, alternative="two-sided").pvalue if len(a) and len(b) else np.nan
            except ValueError:
                p = np.nan
            row[f"MWU {names[i]} vs {names[j]} p"] = p
    results.append(row)

numeric_results = pd.DataFrame(results)
print("Numeric variables comparison")
display(numeric_results)

# Categorical variables: counts, percentages, chi-square (Fisher for 2x2 when needed)
cat_results = []
for var in categorical_vars:
    col = find_col(var)
    if col is None:
        print(f"Column not found: {var}")
        continue
    long_df = pd.concat(
        [d[[col]].assign(group=g) for g, d in group_dfs.items()], ignore_index=True
    ).dropna(subset=[col])
    ct = pd.crosstab(long_df[col], long_df["group"]).reindex(columns=list(group_dfs), fill_value=0)
    pct = ct.div(ct.sum(axis=0).replace(0, np.nan), axis=1).mul(100).round(1)
    print(f"\n{col}: counts")
    display(ct)
    print(f"{col}: % within group")
    display(pct)

    def test_p(table):
        table = table.loc[table.sum(axis=1) > 0, table.sum(axis=0) > 0]
        if table.shape[0] < 2 or table.shape[1] < 2:
            return np.nan
        if table.shape == (2, 2):
            return stats.fisher_exact(table)[1]
        return stats.chi2_contingency(table)[1]

    p_values = {"overall p": test_p(ct)}
    names = list(group_dfs.keys())
    for i in range(len(names)):
        for j in range(i + 1, len(names)):
            p_values[f"{names[i]} vs {names[j]} p"] = test_p(ct[[names[i], names[j]]])
    for category in ct.index:
        row = {"variable": col, "category": category}
        for g in names:
            row[f"{g} n"] = int(ct.loc[category, g])
            row[f"{g} %"] = pct.loc[category, g]
        row.update(p_values)
        cat_results.append(row)

categorical_results = pd.DataFrame(cat_results)
print("\nCategorical variables comparison")
display(categorical_results)

# Save results
numeric_results.to_csv(os.path.join(output_dir, "longcovid_group_comparison_numeric.csv"), index=False)
categorical_results.to_csv(os.path.join(output_dir, "longcovid_group_comparison_categorical.csv"), index=False)
```
